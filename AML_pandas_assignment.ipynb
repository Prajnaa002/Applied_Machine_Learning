Find:

```python
import pandas as pd

data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya"
    ],

    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE"
    ],

    "Marks": [
        85, 78, 92,
        74, 88, 80
    ]
}

df = pd.DataFrame(data)
print("Original DataFrame:\n", df)

# Average marks by department
avg_marks_by_dept = df.groupby('Department')['Marks'].mean()
print("\nAverage marks by Department:\n", avg_marks_by_dept)

# Maximum marks by department
max_marks_by_dept = df.groupby('Department')['Marks'].max()
print("\nMaximum marks by Department:\n", max_marks_by_dept)

# Number of students in each department
students_per_dept = df['Department'].value_counts()
print("\nNumber of students in each Department:\n", students_per_dept)

# For each department, calculate: Mean, Minimum, Maximum, Count
dept_stats = df.groupby('Department')['Marks'].agg(['mean', 'min', 'max', 'count'])
print("\nDepartment-wise Statistics (Mean, Min, Max, Count):\n", dept_stats)
```

Perform the following analysis:

Sort students by FinalMarks.

Add a performance category:

	≥90 → Excellent
	≥80 → Very Good
	≥70 → Good
	<70 → Needs Improvement

Find the department having the highest average FinalMarks.

```python
import pandas as pd
import numpy as np

data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya",
        "John", "Priya"
    ],

    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE",
        "ME", "ECE"
    ],

    "Attendance": [
        92, 75, 96, 68,
        82, 88, 72, 95
    ],

    "Internal": [
        85, 72, 94, 65,
        80, 86, 70, 91
    ],

    "FinalMarks": [
        88, 74, 96, 62,
        83, 89, 68, 93
    ]
}

df = pd.DataFrame(data)

print("Original Student DataFrame:\n", df)

# Display the first five records.
print("\nFirst five records:\n", df.head(5))

# Find the dimensions of the dataset.
print("\nDimensions of the dataset (rows, columns):", df.shape)

# Find the average final mark.
avg_final_mark = df['FinalMarks'].mean()
print(f"\nAverage Final Mark: {avg_final_mark:.2f}")

# Find the highest-scoring student.
highest_scoring_student = df.loc[df['FinalMarks'].idxmax()]
print("\nHighest-scoring student:\n", highest_scoring_student)

# Find the lowest-scoring student.
lowest_scoring_student = df.loc[df['FinalMarks'].idxmin()]
print("\nLowest-scoring student:\n", lowest_scoring_student)

# Display students scoring above 80.
students_above_80 = df[df['FinalMarks'] > 80]
print("\nStudents scoring above 80 in FinalMarks:\n", students_above_80)

# Find average marks by department.
avg_marks_by_dept = df.groupby('Department')['FinalMarks'].mean()
print("\nAverage Final Marks by Department:\n", avg_marks_by_dept)

# Find students having attendance above 80%.
students_high_attendance = df[df['Attendance'] > 80]
print("\nStudents with Attendance above 80%:\n", students_high_attendance)

# Sort students by FinalMarks.
df_sorted_by_final_marks = df.sort_values(by='FinalMarks', ascending=False)
print("\nStudents sorted by FinalMarks (Descending):\n", df_sorted_by_final_marks)

# Add a Result column: Pass if FinalMarks ≥ 70, otherwise Fail.
df['Result'] = np.where(df['FinalMarks'] >= 70, 'Pass', 'Fail')
print("\nDataFrame with 'Result' column:\n", df)

# Add a performance category:
# ≥90 → Excellent, ≥80 → Very Good, ≥70 → Good, <70 → Needs Improvement
conditions = [
    (df['FinalMarks'] >= 90),
    (df['FinalMarks'] >= 80),
    (df['FinalMarks'] >= 70)
]
choices = ['Excellent', 'Very Good', 'Good']
df['PerformanceCategory'] = np.select(conditions, choices, default='Needs Improvement')
print("\nDataFrame with 'PerformanceCategory' column:\n", df)

# Find the department having the highest average FinalMarks.
department_highest_avg_marks = avg_marks_by_dept.idxmax()
highest_avg_marks_value = avg_marks_by_dept.max()
print(f"\nDepartment with the highest average FinalMarks: {department_highest_avg_marks} (Average: {highest_avg_marks_value:.2f})")
```

20. Create:

employees = pd.DataFrame({
    "Employee": [
        "E01", "E02", "E03", "E04",
        "E05", "E06", "E07", "E08"
    ],

    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "Finance"
    ],

    "Experience": [
        5, 3, 7, 4,
        6, 8, 2, 5
    ],

    "Salary": [
        60000, 45000, 75000, 65000,
        55000, 85000, 48000, 70000
    ]
})


Determine:

Average salary.

Maximum and minimum salary.

Employee receiving the highest salary.

Employees earning above ₹60,000.

Average salary for each department.

Average experience by department.

Number of employees in each department.

Sort employees by salary.

Find IT employees with more than 3 years of experience.

Add a SalaryLevel column:

	≥ ₹75,000 → High
	₹50,000–₹74,999 → Medium
	< ₹50,000 → Low

```python
import pandas as pd
import numpy as np

employees = pd.DataFrame({
    "Employee": [
        "E01", "E02", "E03", "E04",
        "E05", "E06", "E07", "E08"
    ],

    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "Finance"
    ],

    "Experience": [
        5, 3, 7, 4,
        6, 8, 2, 5
    ],

    "Salary": [
        60000, 45000, 75000, 65000,
        55000, 85000, 48000, 70000
    ]
})

print("Original Employees DataFrame:\n", employees)

# Determine Average salary.
avg_salary = employees['Salary'].mean()
print(f"\nAverage Salary: ₹{avg_salary:,.2f}")

# Maximum and minimum salary.
max_salary = employees['Salary'].max()
min_salary = employees['Salary'].min()
print(f"Maximum Salary: ₹{max_salary:,.2f}")
print(f"Minimum Salary: ₹{min_salary:,.2f}")

# Employee receiving the highest salary.
highest_paid_employee = employees.loc[employees['Salary'].idxmax()]
print("\nEmployee receiving the highest salary:\n", highest_paid_employee)

# Employees earning above ₹60,000.
employees_above_60k = employees[employees['Salary'] > 60000]
print("\nEmployees earning above ₹60,000:\n", employees_above_60k)

# Average salary for each department.
avg_salary_by_dept = employees.groupby('Department')['Salary'].mean()
print("\nAverage Salary for each Department:\n", avg_salary_by_dept)

# Average experience by department.
avg_experience_by_dept = employees.groupby('Department')['Experience'].mean()
print("\nAverage Experience by Department:\n", avg_experience_by_dept)

# Number of employees in each department.
employees_per_dept = employees['Department'].value_counts()
print("\nNumber of Employees in each Department:\n", employees_per_dept)

# Sort employees by salary.
employees_sorted_by_salary = employees.sort_values(by='Salary', ascending=False)
print("\nEmployees sorted by Salary (Descending):\n", employees_sorted_by_salary)

# Find IT employees with more than 3 years of experience.
it_employees_experienced = employees[
    (employees['Department'] == 'IT') & (employees['Experience'] > 3)
]
print("\nIT Employees with more than 3 years of experience:\n", it_employees_experienced)

# Add a SalaryLevel column:
# ≥ ₹75,000 → High
# ₹50,000–₹74,999 → Medium
# < ₹50,000 → Low
conditions = [
    (employees['Salary'] >= 75000),
    (employees['Salary'] >= 50000) & (employees['Salary'] < 75000)
]
choices = ['High', 'Medium']
employees['SalaryLevel'] = np.select(conditions, choices, default='Low')
print("\nDataFrame with 'SalaryLevel' column:\n", employees)
```
