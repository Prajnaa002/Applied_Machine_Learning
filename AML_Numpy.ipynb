Using NumPy boolean indexing, find:

Marks greater than 75
Marks between 50 and 80

Percentage of students who scored 60 or above

Do not use a Python for loop.

```python
marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84])
# marks greater than 75
greater_than_75 = marks>75
print(marks[greater_than_75])
# marks between 50 and 80
between_50_and_80 = (marks>50) & (marks<80)
print(marks[between_50_and_80])
#Number of students who scored below 50
scored_below_50 = marks>50
print( "Marks scored below 50: "+ str(marks[scored_below_50]))
print("Number of students who scored below 50 : " + str((len(marks[scored_below_50]))))
#Percentage of students who scored 60 or above
scored_above_60 = marks>60
print( "Marks scored above 60: "+ str(marks[scored_above_60]))
print("Number of students who scored above 60 : " + str((len(marks[scored_above_60]))))
percentage = (len(marks[scored_above_60])/len(marks))*100
print("Percentage of students who scored 60 or above : " + str(percentage))
```

12. Daily temperatures for two weeks are:

temperature = np.array([
    31, 32, 33, 35, 36, 34, 32,
    30, 29, 31, 33, 35, 37, 36
])

Find:

Mean temperature

Maximum and minimum temperatures

Number of days above the average temperature

Temperatures greater than 34°C

Difference between maximum and minimum

Standard deviation

```python
temperature = np.array([
    31, 32, 33, 35, 36, 34, 32,
    30, 29, 31, 33, 35, 37, 36
])

#Mean temperature
mean_temperature = np.mean(temperature)
print("Mean temperature:", mean_temperature)

#Maximum and minimum temperatures
max_temperature = np.max(temperature)
min_temperature = np.min(temperature)
print("Maximum temperature:", max_temperature)
print("Minimum temperature:", min_temperature)

#Number of days above the average temperature
days_above_average = np.sum(temperature > mean_temperature)
print("Number of days above the average temperature:", days_above_average)

#Temperatures greater than 34°C
temperature_above_34 = temperature>34
print( "Temperatures greater than 34°C : "+ str(temperature[temperature_above_34]))

#Difference between maximum and minimum
temperature_difference = max_temperature - min_temperature
print("Difference between maximum and minimum:", temperature_difference)

#Standard deviation
temperature_std = np.std(temperature)
print("Standard Deviation:", temperature_std)
```

14. Consider marks of five students in four subjects:

marks = np.array([
    [78, 85, 90, 88],
    [65, 70, 72, 68],
    [92, 88, 95, 91],
    [55, 60, 58, 62],
    [80, 75, 82, 79]
])

Write a NumPy program that calculates:

Total marks of every student

Average of every student

Average mark for every subject

Highest-scoring student

Lowest-scoring student

Students whose average is greater than 80

Overall class average

Then assign a performance category based on average:

|  Average | Category          |
| -------: | ----------------- |
|     ≥ 85 | Excellent         |
| 70–84.99 | Good              |
| 50–69.99 | Average           |
|     < 50 | Needs Improvement |

Challenge: Try assigning these categories using np.where() or np.select() instead of a Python loop.

```python
marks = np.array([
    [78, 85, 90, 88],
    [65, 70, 72, 68],
    [92, 88, 95, 91],
    [55, 60, 58, 62],
    [80, 75, 82, 79]
])

student_total_marks = np.sum(marks, axis=1)
print("Total marks of each student:", student_total_marks)

student_average_marks = np.mean(marks, axis=1)
print("Average marks of each student:", student_average_marks)

subject_average_marks = np.mean(marks, axis=0)
print("Average marks in each subject:", subject_average_marks)

highest_scoring_student_index = np.argmax(student_total_marks)
print("Highest-scoring student (index, 0-based):", highest_scoring_student_index)

lowest_scoring_student_index = np.argmin(student_total_marks)
print("Lowest-scoring student (index, 0-based):", lowest_scoring_student_index)

students_avg_greater_than_80 = marks[student_average_marks > 80]
print("Marks of students whose average is greater than 80:\n", students_avg_greater_than_80)

overall_class_average = np.mean(marks)
print("Overall class average:", overall_class_average)

conditions = [
    student_average_marks >= 85,
    (student_average_marks >= 70) & (student_average_marks < 85),
    (student_average_marks >= 50) & (student_average_marks < 70),
    student_average_marks < 50
]
choices = ['Excellent', 'Good', 'Average', 'Needs Improvement']

performance_categories = np.select(conditions, choices, default='Unknown')
print("\nPerformance categories for each student:", performance_categories)
```

21. Take your own example to normalize an array.



```python
# Take your own example to normalize an array.
# Min-Max Normalization: (x - min(x)) / (max(x) - min(x))

my_array_to_normalize = np.array([10, 20, 30, 40, 50, 100])
print("Original array for normalization:", my_array_to_normalize)

min_val = np.min(my_array_to_normalize)
max_val = np.max(my_array_to_normalize)

normalized_array = (my_array_to_normalize - min_val) / (max_val - min_val)
print("Normalized array:", normalized_array)
```

22. use your own data to standardize an array.

```python
#Use your own data to standardize an array.
# Standardization: (x - mean(x)) / std(x)

my_array_to_standardize = np.array([5, 10, 15, 20, 25, 30])
print("Original array for standardization:", my_array_to_standardize)

mean_val = np.mean(my_array_to_standardize)
std_dev = np.std(my_array_to_standardize)

standardized_array = (my_array_to_standardize - mean_val) / std_dev
print("Standardized array:", standardized_array)
```

24. Compare the time required to multiply 1,000,000 numbers using a Python list and a NumPy array.


```python
import time

SIZE = 1_000_000

# Python list multiplication
python_list1 = list(range(SIZE))
python_list2 = list(range(SIZE))

start_time = time.time()
python_result = [a * b for a, b in zip(python_list1, python_list2)]
end_time = time.time()
python_time = end_time - start_time
print(f"Time taken for Python list multiplication: {python_time:.6f} seconds")

# NumPy array multiplication
numpy_array1 = np.arange(SIZE)
numpy_array2 = np.arange(SIZE)

start_time = time.time()
numpy_result = numpy_array1 * numpy_array2
end_time = time.time()
numpy_time = end_time - start_time
print(f"Time taken for NumPy array multiplication: {numpy_time:.6f} seconds")

print(f"\nNumPy is approximately {python_time / numpy_time:.2f} times faster than Python lists for this operation.")
```
