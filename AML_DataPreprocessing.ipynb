```python
#split independant variables
X = dataset.iloc[:,:-1].values
X
```

```python
#split dependent variable
y = dataset.iloc[:,3].values
y
```

```python
#Taking care of missing data
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(missing_values = np.nan, strategy = 'mean') #NaN/None #Object of Simple Imputer class
imputer = imputer.fit(X[:,1:3]) # used to calculat the mean value for columns 1 and 3
X[:,1:3] = imputer.transform(X[:,1:3]) #fills the missing value
X
```

```python
#encoding categorical data
#encoding the independant variable
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
labelencoder_X = LabelEncoder()
X[:,0] = labelencoder_X.fit_transform(X[:,0]) #X is the independant variable for all rows of column 0 = converts categories into numbers and transform(keep it in the same place)
X
```

```python
#encoding the independant variable
from sklearn.compose import ColumnTransformer
ct = ColumnTransformer([("Country",OneHotEncoder(),[0])], remainder ="passthrough")
x= ct.fit_transform(X)

x.astype(int)
```

```python
#encoding the dependant variable
labelencoder_y = LabelEncoder()
y = labelencoder_y.fit_transform(y)
y
```

```python
#split data into training set and test set
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,random_state=45) #training: 80% and test/validation:20%, random state: what if random state is set to 0?
#X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2) check ?
```
